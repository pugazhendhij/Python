What is a Typecode? </br>
      A typecode is a single character passed as the first argument to array.array(). It acts as a blueprint, telling Python:

1. What data type the array accepts (integers, floating-point numbers, characters).

2. How much memory (bytes) to allocate for each slot in computer memory.

### Memory as Digital Boxes

Think of memory as a row of small boxes that can only hold a fixed number of digits (like a small digital counter on an old clock).

### Unsigned Integer: Positive Only

* **What it means:** It cannot hold negative numbers. It uses all its storage space to count upwards starting from $0$.
* **Real-world analogy:** A digital step tracker. You can walk $0$ steps or $10,000$ steps, but you can never walk $-50$ steps.
* **8-bit example:** Can store numbers from $0$ to $255$.

### Signed Integer: Positive AND Negative

* **What it means:** It splits its storage capacity so it can hold both positive and negative numbers.
* **Real-world analogy:** A thermometer. Temperature can go above zero ($+30^\circ\text{C}$) or drop below zero ($-10^\circ\text{C}$).
* **8-bit example:** Can store numbers from $-128$ to $+127$.
