### Shallow Copy vs. Deep Copy

* **Shallow Copy:** Creates a new outer array, but it shares references to any nested elements inside it. Changes to inner lists will affect both the original and copied arrays.
* **Deep Copy:** Recursively duplicates every layer of the array and its nested elements into entirely new memory locations. Both copies are completely independent, so modifications to one will never alter the other.

---

| Type | Outer Array | Nested Elements | Independence |
| --- | --- | --- | --- |
| **Shallow Copy** | New memory reference | Shared references | Interdependent (inner elements) |
| **Deep Copy** | New memory reference | New memory references | Completely independent |

```python
## Shallow Copy like Assignment Operator

import array as arr

data = arr.array('f',[11,12,13,14,15])

a = data

print(f'Original Array Value{data}')

print(id(a),id(data))
```

```python

## DeepCopy
import array as arr
import copy

data = arr.array('f',[11,12,13,14,15])

a = copy.deepcopy(data)

print(f'Original Array Value{data}')

print(id(a),id(data))
```
